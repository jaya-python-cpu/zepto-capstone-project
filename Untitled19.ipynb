```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import sqlite3
import re
from urllib.parse import urljoin
#1. SCRAPE 60 BOOKS
urls = ["https://books.toscrape.com/","https://books.toscrape.com/catalogue/page-2.html","https://books.toscrape.com/catalogue/page-3.html"]
books_all = []
for url in urls:
  r = requests.get(url)
  soup = BeautifulSoup(r.text, "html.parser")
  books_all.extend(soup.select("article.product_pod"))
print("TOTAL BOOKS:",len(books_all))
# CLEAN DATA
rating_map = {"One": 1, "Two": 2, "Three": 3, "Four": 4, "Five": 5}
for book in books_all:
  title = book.h3.a["title"]
  price_text = book.select_one(".price_color").text
  rating_text = book.select_one(".star-rating")["class"][1]
  availability = book.select_one(".availability").text.strip()
  match = re.search(r"\d+(\.\d+)?", price_text)
  price_gbp = float(match.group()) if match else None
  books_data.append({"title": title, "price_gbp": price_gbp, "rating": rating_map[rating_text], "availability": availability,"in_stock": availability.lower() == "in stock", "price_inr":price_gbp *105.50 if price_gbp else None})
df = pd.DataFrame(books_data)
# 3. CATEGORY
categories = []
for book in books_all:
    try:
     link = urljoin("https://books.toscrape.com/", book.h3.a["href"])
     page = BeautifulSoup(requests.get(link).text, "html.parser")
     crumbs = page.select("ul.breadcrumb li")
     if len (crumbs) >= 3:
        category = crumbs[2].get_text(strip=True)
     else:
         category = "unknown"
    except Exeption:
        category = "unknown"
    categories.append(category)
df["category"] = categories
print("Rows:", len(df))
print("CATAGORIES:", df["category"].nunique())
  # 4. SQLITE DATABASE
      conn = sqlite3.connect("zepto_books.db")
      conn.execute("PRAGMA FOREIGN_KEYS = ON")
      conn.execute("DROP TABLE IF EXISTS books")
      conn.execute("DROP TABLE IF EXISTS categories")
      conn.execute("""CREATE TABLE categories(category_id INTEGER PRIMARY KEY AUTOINCREMENT,category_name TEXT UNIQUE)""")
  for cat in df["category"].unique():
      conn.execute("""INSERT INTO categories(category_name) VALUES(?)""",(cat,))
      conn.execute("""CREATE TABLE books(book_id INTEGER PRIMARY KEY AUTOINCREMENT,title TEXT,price_gbp REAL, price_inr REAL,rating INTEGER,availability TEXT, in_stock INTEGER,category_id INTEGER,FOREIGN KEY(category_id) REFERENCES categories(category_id))""")
  for _, row in df.iterrows():
      category_id = conn.execute("SELECT category_id FROM categories WHERE category_name=?",(row["category"],)).fetchone()[0]
      conn.execute("""INSERT INTO books(title,price_gbp,price_inr,rating,availability,in_stock,category_id) VALUES(?,?,?,?,?,?,?)""",(row["title"],row["price_gbp"],row["rating"],row["availability"],row["in_stock"],category_id))
      conn.commit()
  # 5. SQL QUERIES
  q1 = pd.read_sql("""SELECT title, price_inr FROM books ORDER BY price_inr DESC LIMIT 10""", conn)
  q2 = pd.read_sql("""SELECT DISTINCT rating FROM books ORDER BY rating""", conn)
  q3 = pd.read_sql("""SELECT title, price_inr FROM books WHERE price_inr BETWEEN 1000 AND 5000""", conn)
  q4 = pd.read_sql("""SELECT title, rating FROM books WHERE rating IN (4,5)""", conn)
  q5 = pd.read_sql("""SELECT b.title, b.price_inr, c.category_name FROM books b JOIN categories c ON b.category_id = c.category_id LIMIT 10""", conn)
  # 6. PANDAS JOIN
  books_sql = pd.read_sql("SELECT * FROM books", conn)
  categories_sql = pd.read_sql("SELECT * FROM categories", conn)
  pandas_join = books_sql.merge(categories_sql, on="category_id", how="inner")
  # 7. SAVE CSV
  df.to_csv("zepto_books_cleaned.csv", index=False)
  print("SQL QUERIES COMLEATED")
  print("BOOK ROWS:", len(books_sql))
  print("CATEGORY ROWS:", len(categories_sql))
  print("PANDAS JOIN ROWS:", len(pandas_join))
  print("MODULE 1 CORE WORK COMLEATED")

```
